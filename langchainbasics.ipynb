### MiniProject 1: Simple LLM Call with Streaming

```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage,SystemMessage
```

```python
model=init_chat_model("groq:openai/gpt-oss-120b")
model
```

```python
#other way to load the model
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI
llm=ChatGroq(model="openai/gpt-oss-120b")
llm
```

```python
## Create messages
messages=[
    SystemMessage("You are a helpful AI assistant"),
    HumanMessage("What is Machine Learning and LLM?")
]
```

```python
## invoke the model response
response=model.invoke(messages)
response
```

```python
print(response.content)
```

```python
model.invoke([HumanMessage("What is machine learning")])
```

```python
## streaming example - the content starts priniting as it is returned
for chunk in model.stream(messages):
    print(chunk.content, end="", flush=True)
```

### MiniProject 2 - Dynamic Prompt Templates

```python
from langchain_core.prompts import ChatPromptTemplate

## create translation app

translation_template=ChatPromptTemplate.from_messages([
    ("system","You are a professional translator. Translate the follow text {text} from {source_language} to {target_language}. Maintain the tone and style"),
    ("user","{text}")
])

## using the template
prompt=translation_template.invoke({
    "source_language":"English",
    "target_language":"German",
    "text":"Langchain makes building AI application incredibly easy!"
})
prompt
```

```python
for chunk in model.stream(prompt):
    print(chunk.content, end="", flush=True)
print()
translated_response=model.invoke(prompt)
print(translated_response.content)
print()
```

### MiniProject 3: Building your First Chain

```python
from langchain_core.output_parsers import StrOutputParser   #useful to bring string outputs from llms, doesnt take that ai message ... from output
from langchain_core.runnables import RunnablePassthrough, RunnableLambda

# create a complex chain
def create_story_chain():
    # template for story generation
    story_prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a creative storyteller. Write a short, engaging story based on the give theme, character and setting."),
        ("user","Theme: {theme} \n Main character: {character} \n Setting: {setting}")
    ])

    # template for story analysis
    analysis_prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a literary critic. Analyze the following story and provide insights."),
        ("user", "{story}")
    ])

    # build the chain - method 1: sequential execution
    story_chain = (
        story_prompt 
        | model 
        | StrOutputParser()
    )
    # create a small function to pass the story to analysis
    def analyze_story(story_text):
        return {"story": story_text}

    analysis_chain = (
        story_chain
        | RunnableLambda(analyze_story)
        | analysis_prompt
        | model
        | StrOutputParser()
    )
    return analysis_chain
```

```python
chain=create_story_chain()
chain
```

```python
result = chain.invoke({
    "theme": "artificial intelligence",
    "character": "a curious robot",
    "setting": "a futuristic city"
})

print("Story and Analysis:")
print(result)
```

```python
result = chain.invoke({
    "theme": "epic adventure",
    "character": "Azhor Ahai, light bringer who will vanquish all darkness",
    "setting": "medieval high fantasy"
})

print("Story and Analysis:")
print(result)
```
